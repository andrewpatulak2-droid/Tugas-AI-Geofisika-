### Penjelasan Elemen Kerangka R–T–C–I–K–O–Q–V

* **R (Role):** Posisi atau sudut pandang keahlian yang diemban oleh kecerdasan buatan.
* **T (Task):** Sasaran pekerjaan utama yang wajib diselesaikan sistem.
* **C (Context):** Latar belakang permasalahan serta skenario di balik studi kasus.
* **I (Input):** Kumpulan parameter, data, atau variabel awal yang dilampirkan.
* **K (Constraints):** Ketentuan pembatas, larangan, serta aturan main operasional.
* **O (Output):** Bentuk susunan dan format laporan hasil akhir yang diinginkan.
* **Q (Quality Criteria):** Standar mutu dan tolak ukur keberhasilan jawaban.
* **V (Verification):** Langkah-langkah pengujian kebenaran dan validasi hasil.

### Prompt 1 — Pemodelan Interpretasi Seismik Refraksi

* **R — Role:** Posisikan diri Anda sebagai seorang konsultan senior geofisika bidang refraksi yang bertugas mendampingi peserta didik program sarjana dalam mengkaji catatan waktu penjalaran gelombang elastik.
* **T — Task:** Lakukan analisis mendalam terhadap parameter waktu kedatangan awal (first arrival) untuk mengidentifikasi profil kecepatan rambat tiap lapisan tanah serta menaksir kedalaman bidang batas litologi di bawah permukaan.
* **C — Context:** Teknik survei seismik dangkal sering diaplikasikan untuk memetakan struktur geologi dekat permukaan, contohnya pemisahan antara zona pelapukan dengan batuan dasar yang masif. Tugas ini merupakan bagian dari modul praktikum Metode Seismik Refraksi guna melatih keterampilan pembacaan grafik waktu terhadap jarak.
* **I — Input:** Manfaatkan lembar data yang mencakup variabel jarak sensor (geofon) terhadap titik ledakan (offset) serta catatan waktu tiba gelombang pertama. Harap cermati kesesuaian satuan waktu dan jarak yang disertakan.
* **K — Constraints:** Jaga keaslian angka-angka data tanpa melakukan modifikasi sepihak. Uraikan secara sistematis cara memilah titik-titik kurva menjadi tren gelombang langsung dan bias. Terangkan asumsi dasar yang digunakan, misalnya model bidang datar dengan kecepatan yang bertambah seiring kedalaman. Jika menjumpai data yang menyimpang dari tren umum, berikan ulasan analitis tanpa langsung membuangnya. Hindari penarikan vonis jenis batuan secara mutlak hanya berpatokan pada besaran kecepatan gelombang.
* **O — Output:** Format laporan wajib menyajikan penjelasan ringkas karakteristik dataset dan satuan ukur, prosedur pembacaan diagram waktu-jarak beserta pemisahan segmennya, besaran kecepatan gelombang per lapisan beserta metode penurunannya, perhitungan estimasi ketebalan stratum pertama beserta rumus yang dipakai, narasi deskriptif penafsiran kondisi geologi secara objektif, serta evaluasi keterbatasan metode dan ringkasan simpulan.
* **Q — Quality Criteria:** Seluruh proses hitungan wajib akurat dengan konsistensi satuan yang terjaga, logika alur penyelesaian mudah diikuti mahasiswa strata satu, interpretasi tidak melenceng dari bukti data, serta format dokumen siap lampir untuk tugas praktikum.
* **V — Verification:** Pastikan kecepatan lapisan bawah bernilai lebih tinggi dari lapisan di atasnya, periksa apakah titik potong kurva (crossover) sudah konsisten, dan uji kewajaran hasil akhir dibandingkan kisaran kecepatan material di alam.

### Prompt 2 — Evaluasi Statistik Katalog Gempabumi

* **R — Role:** Bertindaklah sebagai periset seismologi yang membimbing mahasiswa tingkat sarjana untuk mengulik pola dan karakteristik dari basis data kejadian gempa.
* **T — Task:** Lakukan kajian statistik terhadap katalog gempa untuk mengeksplorasi sebaran geografis episenter, variasi kedalaman hiposenter, tren magnitudo, serta periodisitas waktu kejadian di zona penelitian.
* **C — Context:** Arsip katalog gempabumi merekam parameter waktu, koordinat titik pusat, kedalaman, dan kekuatan gempa. Analisis ini dijalankan dalam mata kuliah Seismologi untuk memahami dinamika kegempaan regional serta korelasinya dengan jalur tektonik aktif maupun zona subduksi.
* **I — Input:** Gunakan dokumen katalog gempa yang memuat informasi waktu, posisi koordinat, kedalaman, serta skala magnitudo. Batasi seluruh proses pengolahan hanya berdasarkan isi data tersebut.
* **K — Constraints:** Dilarang mengubah nilai parameter asli dalam katalog. Jangan mencampuradukkan tipe skala magnitudo yang berbeda tanpa penjabaran. Berikan penjelasan mengenai penanganan baris data yang kosong atau berada di luar area kajian. Dilarang membuat estimasi prediksi bencana gempa atau analisis tingkat risiko wilayah. Kaitan dengan kerangka tektonik lokal hanya boleh dipaparkan sebagai hipotesis ilmiah yang beralasan.
* **O — Output:** Hasil analisis harus mencakup ringkasan deskriptif katalog meliputi rentang waktu, batasan wilayah, dan jumlah kejadian, analisis parameter statistik dasar magnitudo dan kedalaman, pola persebaran spasial titik episenter beserta profil kedalamannya, grafik tren kemunculan peristiwa gempa terhadap kronologi waktu, uraian interpretasi hasil dalam bentuk paragraf naratif, serta poin kesimpulan singkat.
* **Q — Quality Criteria:** Nilai numerik harus sinkron dengan katalog, penggunaan peristilahan seismologi tepat secara ilmiah, interpretasi tektonik tidak berlebihan, dan laporan memenuhi standar tugas akademik Seismologi.
* **V — Verification:** Validasi ulang jumlah total event yang diteliti, pastikan rentang magnitudo serta kedalaman dilaporkan secara presisi, dan pastikan setiap kesimpulan terhubung langsung dengan rekam jejak data katalog.

### Prompt 3 — Analisis Geostatistika dan Pemetaan Spasial

* **R — Role:** Ambil peran sebagai ahli geostatistika yang membimbing mahasiswa sarjana Geofisika dalam mengkaji struktur ketergantungan spasial serta teknik interpolasi parameter.
* **T — Task:** Selidiki pola korelasi jarak antar titik pengamatan memakai pendekatan variogram, kemudian jabarkan pemanfaatannya dalam metode estimasi kriging serta evaluasi tingkat keandalan petanya.
* **C — Context:** Data pengukuran lapangan geofisika umumnya bersifat titik diskrit, sementara pembuatan peta kontur menghendaki informasi yang menyeluruh pada setiap area. Topik ini dibahas pada mata kuliah Geostatistika & Analisis Spasial guna memahami derajat kemiripan antar lokasi berdekatan sebelum melakukan penaksiran spasial.
* **I — Input:** Manfaatkan dataset berupa titik koordinat pengamatan beserta satu atribut variabel fisika batuan, misalnya anomali gayaberat, ketebalan lapisan sedimen, atau elevasi muka air tanah. Perhatikan sistem referensi koordinat serta satuan panjang data.
* **K — Constraints:** Pertahankan nilai ukur asli tanpa intervensi perubahan yang tidak berdasar. Pastikan sistem koordinat sudah dikonversi ke satuan meter sebelum perhitungan jarak spasial. Jangan menghapus data anomali (outlier) secara sembunyi-sembunyi; jelaskan dampak keberadaannya. Berikan argumen rasional di balik pemilihan jarak jeda antar pasangan titik (lag size) dan model variogram teoretis. Hindari klaim akurasi tinggi untuk estimasi titik di luar sebaran data observasi.
* **O — Output:** Laporan akhir wajib menampilkan deskripsi karakteristik dataset beserta parameter statistik deskriptif, penjabaran bentuk variogram eksperimental dan cara membacanya, arti fisis dari komponen nugget, sill, dan range pada kasus ini, penentuan model teoretis variogram beserta alasan pemilihan tipenya, uraian peta hasil kriging beserta evaluasi tingkat ketidakpastian spasialnya, komparasi ringkas dengan metode interpolasi deterministik konvensional, dan kesimpulan menyeluruh.
* **Q — Quality Criteria:** Konsep geostatistik dijabarkan secara runtut dan komunikatif, seluruh nilai parameter disertakan satuannya, aspek keterbatasan jumlah titik sampel dibahas terbuka, dan hasil siap diaplikasikan dalam laporan.
* **V — Verification:** Uji apakah nilai range logis terhadap bentang wilayah studi, pastikan validasi silang (seperti teknik leave-one-out) diterapkan pada model, dan cek apakah simpulan analisis selaras dengan hasil validasi tersebut.

### Prompt 4 — Simulasi Model Termal Litosfer Samudera

* **R — Role:** Berperanlah sebagai pakar geodinamika dan pemodelan numerik yang mendampingi mahasiswa S1 Geofisika untuk memahami proses pendinginan termal litosfer samudera.
* **T — Task:** Kaji dan bedah formulasi matematika model pendinginan konduktif satu dimensi, serta bandingkan hasil simulasi numerik terhadap solusi analitik eksaknya pada berbagai variasi usia lempeng.
* **C — Context:** Litosfer samudera mengalami penciutan dan penambahan ketebalan seiring bertambahnya usia akibat hilangnya fluks panas ke arah permukaan bumi. Kajian ini dipelajari dalam mata kuliah Geodinamika, Tektonofisika, dan Geofisika Komputasi untuk menghubungkan usia lempeng dengan profil suhu vertikal.
* **I — Input:** Gunakan parameter tetapan yang ditentukan, seperti koefisien difusivitas termal, temperatur permukaan, suhu mantel bawah, kedalaman batas maksimum, serta beberapa variasi umur lempeng tektonik. Larangan keras menggunakan variabel luar tanpa izin.
* **K — Constraints:** Parameter masukan tidak boleh diubah kecuali dicatat secara transparan. Jelaskan batasan asumsi model, termasuk dinamika yang diabaikan seperti peluruhan panas internal dan aliran konveksi. Pastikan keselarasan satuan di semua persamaan, khususnya konversi umur lempeng dalam jutaan tahun. Uraikan syarat kestabilan komputasi numerik yang harus dipatuhi. Jangan langsung menyamakan hasil model teoretis sederhana ini dengan kondisi nyata interior bumi yang kompleks.
* **O — Output:** Format luaran harus memuat penjelasan persamaan pengatur dan asumsi fisik dengan bahasa yang mudah dipahami, penentuan parameter kondisi awal serta kondisi batas sistem, grafik perbandingan profil temperatur terhadap kedalaman untuk tiap kelompok usia, analisis komparasi hasil numerik vs analitik beserta besaran deviasinya, penentuan posisi kedalaman isoterm acuan dan korelasinya terhadap umur lempeng, uraian interpretasi implikasi geodinamika dalam bentuk paragraf, serta evaluasi batasan model dan simpulan ringkas.
* **Q — Quality Criteria:** Prinsip dasar fisika termal dijelaskan dengan benar, selisih perhitungan numerik dan analitik dijabarkan dalam angka kuantitatif, konsistensi satuan terjaga, serta interpretasi tidak keluar dari batas asumsi model.
* **V — Verification:** Periksa apakah deviasi hasil numerik semakin kecil ketika resolusi kedalaman grid diperhalus, pastikan kondisi batas suhu permukaan dan dasar terpenuhi, serta cek apakah posisi isoterm semakin menunjam pada lempeng yang lebih tua.

### Prompt 5 — Klasifikasi Litologi Sumur Menggunakan Pembelajaran Mesin

* **R — Role:** Bertindaklah sebagai spesialis kecerdasan artifisial geosains yang membimbing mahasiswa sarjana Geofisika dalam menerapkan algoritma machine learning pada data sumur pemboran.
* **T — Task:** Rancang, terapkan, dan evaluasi performa model pembelajaran mesin guna mengidentifikasi kelas litologi formasi berdasarkan data log sumur, lengkap dengan komparasi terhadap model acuan (baseline) sederhana.
* **C — Context:** Kurva rekaman log sumur seperti gamma ray, densitas batuan, porositas neutron, serta kecepatan sonik menyimpan informasi penting untuk membedakan jenis lapisan batuan. Analisis ini dibahas dalam mata kuliah AI untuk Geofisika guna memahami potensi serta kendala penerapan kecerdasan buatan pada interpretasi bawah permukaan.
* **I — Input:** Gunakan dataset latihan yang memuat informasi identitas sumur, kedalaman, kurva log geofisika, serta label kelas litologi acuan. Cermati satuan masing-masing kurva log serta asal-usul validitas label.
* **K — Constraints:** Dilarang memodifikasi nilai data asli tanpa dasar metodologi yang dapat dipertanggungjawabkan. Pembagian data latih dan data uji wajib berbasis pemisahan sumur (well-based split), bukan diacak per baris, dan sertakan argumentasinya. Dilarang menggunakan informasi data uji saat proses penyetelan hyperparameter model. Atribut kedalaman sumur dan identitas nama sumur tidak boleh dijadikan sebagai fitur input pemodelan. Jelaskan mekanisme penanganan data kosong atau nilai atribut yang tidak logis secara fisika batuan. Hindari klaim bahwa model cerdas ini bisa langsung digeneralisasi secara universal ke lapangan migas lain.
* **O — Output:** Laporan analisis wajib menyajikan ringkasan deskripsi dataset beserta komposisi sebaran kelas litologi, penjelasan arsitektur model pembanding sederhana dan model machine learning utama, evaluasi performa yang merinci metrik tiap kelas litologi, bukan hanya persentase akurasi total, analisis kesalahan (error analysis) mengenai litologi yang paling sering tertukar beserta tinjauan alasannya secara geologis, pembahasan pengaruh ketidakseimbangan jumlah sampel antar kelas (class imbalance), serta uraian interpretasi hasil dalam bentuk narasi paragraf dan kesimpulan padat.
* **Q — Quality Criteria:** Evaluasi model harus objektif dan bebas dari kebocoran informasi (no data leakage), performa dijabarkan secara rinci per kelas, penjelasan mudah dipahami level sarjana, serta hasil dapat langsung disematkan ke laporan tugas.
* **V — Verification:** Pastikan tidak ada data sumur yang sama muncul di dalam data latih sekaligus data uji, verifikasi bahwa algoritma machine learning terbukti lebih unggul dari model baseline, dan pastikan performa anjlok ke tingkat acak apabila label kelas sengaja di-shuffling.
